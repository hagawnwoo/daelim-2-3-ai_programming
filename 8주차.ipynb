```python
import pandas as pd

df = pd.DataFrame(student_data)

df["name"] = df["name"].str.strip()

df["email"] = df["email"].str.strip().str.lower()

print(df[["name", "email"]])
```

```python
df["domain"] = df["email"].apply(lambda x: x.split("@")[1] if "@" in x else  "invalid")

print(df[["email", "domain"]])
```

```python
df["final_score"] = df[["assignment_score", "quiz_avg", "project_score"]].mean(axis=1)

print(df[["assignment_score", "quiz_avg", "project_score", "final_score"]])
```

```python
df["final_score"] = df["final_score"].apply(lambda x: str(round(x, 1)))

print(df[["final_score"]])
```

```python
df["achievement_level"] = df["final_score"].apply(lambda x: 
    "우수" if float(x) >= 85 else
    "보통" if float(x) >= 70 else
    "주의"
)

print(df[["final_score", "achievement_level"]])
```
